```python
print("1. What is the average, median, and mode of PurchaseAmount?")
print("Average =", df['PurchaseAmount'].mean())
print("Median =", df['PurchaseAmount'].median())
print("Mode =", df['PurchaseAmount'].mode()[0])
```

```python
print("2.Are there any outliers in the PurchaseAmount data")
Q1 = df['PurchaseAmount'].quantile(0.25)
Q3 = df['PurchaseAmount'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['PurchaseAmount'] < lower) |
              (df['PurchaseAmount'] > upper)]
print("Number of Outliers =", len(outliers))
```

```python
from scipy import stats
print("3. Skewness and Kurtosis")
print("Skewness =", stats.skew(df['PurchaseAmount']))
print("Kurtosis =", stats.kurtosis(df['PurchaseAmount']))
```

```python
from scipy import stats
print("4. Is there a significant difference in spending between male and female customers?")
male = df[df['Gender'] == 'Male']['PurchaseAmount']
female = df[df['Gender'] == 'Female']['PurchaseAmount']
t_stat, p_value = stats.ttest_ind(male, female)
print("P Value =", p_value)
if p_value < 0.05:
    print("There is a significant difference.")
else:
    print("There is no significant difference.")
```

```python
from scipy.stats import chi2_contingency
print("5. Is there a relationship between ProductCategory and Churn?")
table = pd.crosstab(df['ProductCategory'], df['Churn'])
chi2, p, dof, expected = chi2_contingency(table)
print("P Value =", p)
if p < 0.05:
    print("There is a relationship.")
else:
    print("There is no relationship.")
```

```python
from scipy.stats import f_oneway
print("6.Does PurchaseAmount vary across different regions?")
regions = df['Region'].unique()
groups = [df[df['Region']==r]['PurchaseAmount'] for r in regions]
f_stat, p_value = f_oneway(*groups)
print("P Value =", p_value)
if p_value < 0.05:
    print("Purchase amount varies across regions.")
else:
    print("Purchase amount does not vary across regions.")
```

```python
print("7. Best Campaign")
campaign = df.groupby('CampaignGroup')['PurchaseAmount'].mean()
print(campaign)
print("Best Campaign =", campaign.idxmax())
```

```python
from scipy import stats
print("8.Does PurchaseAmount follow a normal distribution")
sample = df['PurchaseAmount'].sample(500)
stat, p_value = stats.shapiro(sample)
print("P Value =", p_value)
if p_value > 0.05:
    print("Data follows normal distribution.")
else:
    print("Data does not follow normal distribution.")
```

```python
print("9. Central Limit Theorem")
if len(df) >= 30:
    print("CLT is applicable because sample size is greater than 30.")
else:
    print("CLT is not applicable.")
```

```python
import numpy as np
from scipy import stats
print("10. What is the 95% Confidence Interval?")
mean = df['PurchaseAmount'].mean()
std = df['PurchaseAmount'].std()
n = len(df)
margin = stats.t.ppf(0.975, n-1) * (std / np.sqrt(n))
lower = mean - margin
upper = mean + margin
print("95% Confidence Interval =")
print(lower, "to", upper)
```
